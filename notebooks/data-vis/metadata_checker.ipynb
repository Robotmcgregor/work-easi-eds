# Metadata checker

```python
from osgeo import gdal

fp = "/home/jovyan/scratch/eds/tiles/p089r078/sr/2024/202408/ls89sr_p089r078_20240831_nbart6m6_clr.tif"

ds = gdal.Open(fp)
assert ds is not None, "Failed to open raster"

print("Driver:", ds.GetDriver().ShortName)
print("Raster size:", ds.RasterXSize, ds.RasterYSize)
print("Bands:", ds.RasterCount)

print("\n=== GLOBAL METADATA ===")
meta = ds.GetMetadata()
if not meta:
    print("(no global metadata)")
else:
    for k, v in meta.items():
        print(f"{k}: {v}")
```

```python
print("\n=== METADATA DOMAINS ===")
domains = ds.GetMetadataDomainList()
print(domains)

for d in domains:
    print(f"\n--- Domain: {d} ---")
    md = ds.GetMetadata(d)
    for k, v in md.items():
        print(f"{k}: {v}")
```

```python
for i in range(1, ds.RasterCount + 1):
    band = ds.GetRasterBand(i)
    print(f"\n=== BAND {i} ===")
    print("NoData:", band.GetNoDataValue())

    bmeta = band.GetMetadata()
    if not bmeta:
        print("(no band metadata)")
    else:
        for k, v in bmeta.items():
            print(f"{k}: {v}")
```

```python
for i in range(1, ds.RasterCount + 1):
    band = ds.GetRasterBand(i)
    desc = band.GetDescription()
    if desc:
        print(f"Band {i} description:", desc)
```
