```python
import boto3

bucket = "dcceew-eds-data"
prefix = "AROAZ6PFZYT4B4C7MNRHV:robotmcgregor/eds/optimised/manifests/"

s3 = boto3.client("s3")

response = s3.list_objects_v2(Bucket=bucket, Prefix=prefix)

for obj in response.get("Contents", []):
    key = obj["Key"]
    print("Deleting:", key)
    s3.delete_object(Bucket=bucket, Key=key)

print("Finished deleting manifests")
```
