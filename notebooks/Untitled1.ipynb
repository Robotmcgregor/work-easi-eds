```python
import sqlite3
import pandas as pd
from pathlib import Path
```

```python
DB_PATH = Path("/home/jovyan/work-easi-eds/data copy/eds_database.db")

print(DB_PATH)
print(DB_PATH.exists())
```

```python
conn = sqlite3.connect(DB_PATH)

print("Connected successfully")
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table'
ORDER BY name;
"""

tables_df = pd.read_sql_query(query, conn)

tables_df
```

```python
for table in tables_df['name']:
    print(table)
```

```python
row_counts = []

for table in tables_df['name']:
    try:
        count_query = f"SELECT COUNT(*) AS count FROM {table}"
        count = pd.read_sql_query(count_query, conn).iloc[0]['count']

        row_counts.append({
            'table': table,
            'rows': count
        })

    except Exception as e:
        row_counts.append({
            'table': table,
            'rows': f'ERROR: {e}'
        })

row_counts_df = pd.DataFrame(row_counts)
row_counts_df.sort_values('rows', ascending=False)
```
