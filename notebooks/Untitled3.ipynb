```python
from osgeo import gdal
import numpy as np

files = [
    "/home/jovyan/scratch/eds-work-processing/p089r080/ga0_work/sl9olre_p089r080_20250607_ga0-clr_e32756.tif",
    "/home/jovyan/scratch/eds-work-processing/p089r080/ga0_work/sl8olre_p089r080_20260109_ga0-clr_e32756.tif",
]

bands_to_check = [2, 3, 5, 6]  # 1-based GDAL bands corresponding to ref[1], ref[2], ref[4], ref[5]

for fp in files:
    print(f"\nFILE: {fp}")
    ds = gdal.Open(fp)
    if ds is None:
        print("  FAILED TO OPEN")
        continue

    for b in bands_to_check:
        rb = ds.GetRasterBand(b)
        arr = rb.ReadAsArray().astype("float32")
        nodata = rb.GetNoDataValue()

        finite = np.isfinite(arr)
        valid = arr[finite]

        print(f"  Band {b}")
        print(f"    nodata   : {nodata}")
        print(f"    min      : {np.nanmin(valid) if valid.size else 'NA'}")
        print(f"    max      : {np.nanmax(valid) if valid.size else 'NA'}")
        print(f"    <= -1    : {np.sum(valid <= -1)}")
        print(f"    < 0      : {np.sum(valid < 0)}")
        if nodata is not None:
            print(f"    ==nodata : {np.sum(arr == nodata)}")
```
