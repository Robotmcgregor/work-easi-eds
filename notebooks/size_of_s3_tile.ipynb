# Size of Lsat Tile in storage

```python
import boto3

bucket = "dcceew-eds-data"
prefix = "AROAZ6PFZYT4B4C7MNRHV:robotmcgregor/eds/optimised/tiles/p089r078/ndvi/"

s3 = boto3.client("s3")

total_size = 0
total_objects = 0

paginator = s3.get_paginator("list_objects_v2")
for page in paginator.paginate(Bucket=bucket, Prefix=prefix):
    for obj in page.get("Contents", []):
        total_size += obj["Size"]
        total_objects += 1

print(f"Objects: {total_objects}")
print(f"Bytes  : {total_size}")
print(f"GiB    : {total_size / (1024**3):.2f}")
```
