# Delete EDS Outputs

```python
from pathlib import Path
import shutil

target = Path("/home/jovyan/data/compat/files")

assert target.exists(), f"Path does not exist: {target}"
assert target.is_dir(), f"Not a directory: {target}"

print("About to delete ALL contents of:")
print(" ", target.resolve())

# Safety check: refuse to run if path is too short / dangerous
if len(target.parts) < 5:
    raise RuntimeError("Refusing to delete: path too short / unsafe")

for item in target.iterdir():
    if item.is_dir():
        shutil.rmtree(item)
        print("Deleted directory:", item.name)
    else:
        item.unlink()
        print("Deleted file:", item.name)

print("\n Cleanup complete. Directory preserved.")
```
