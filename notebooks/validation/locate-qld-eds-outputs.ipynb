# Locate QLD run data

```python
import boto3

def list_keys_preview(bucket, prefix="", max_keys=50):
    s3 = boto3.client("s3")

    resp = s3.list_objects_v2(
        Bucket=bucket,
        Prefix=prefix,
        MaxKeys=max_keys
    )

    keys = [obj["Key"] for obj in resp.get("Contents", [])]

    print(f"\nBucket: {bucket}")
    print(f"Prefix: {prefix}")
    print(f"Showing {len(keys)} keys:\n")

    for k in keys:
        print(k)

    return keys
```

```python
# list_keys_preview("dcceew-eds-data")
list_keys_preview("dcceew-rs-data")
```

```python
import boto3

def list_top_level_prefixes(bucket):
    s3 = boto3.client("s3")

    resp = s3.list_objects_v2(
        Bucket=bucket,
        Delimiter="/"
    )

    prefixes = [p["Prefix"] for p in resp.get("CommonPrefixes", [])]

    print(f"\nTop-level prefixes in {bucket}:\n")
    for p in prefixes:
        print(p)

    return prefixes
```

```python
list_top_level_prefixes("dcceew-rs-data")
```

```python
import boto3

def find_keys_containing(bucket, prefix="", text="completed_checked", max_show=100):
    s3 = boto3.client("s3")
    paginator = s3.get_paginator("list_objects_v2")

    matches = []

    for page in paginator.paginate(Bucket=bucket, Prefix=prefix):
        for obj in page.get("Contents", []):
            key = obj["Key"]
            if text.lower() in key.lower():
                matches.append(key)

    print(f"Found {len(matches)} keys containing '{text}' under '{prefix}'\n")

    for k in matches[:max_show]:
        print(k)

    if len(matches) > max_show:
        print(f"\n... and {len(matches) - max_show} more")

    return matches
```

```python
completed_checked_keys = find_keys_containing(
    bucket="dcceew-rs-data",
    prefix="dcceew_202602_run/",
    text="completed_checked"
)
```
