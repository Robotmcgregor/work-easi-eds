```python
import boto3

client = boto3.client("s3")
bucket = "dcceew-rsu-data"

paginator = client.get_paginator("list_objects_v2")

count = 0
for page in paginator.paginate(Bucket=bucket):
    for obj in page.get("Contents", []):
        print(obj["Key"])
        count += 1

print(f"\nTotal files found: {count}")
```
