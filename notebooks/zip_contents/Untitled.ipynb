```python
from pathlib import Path
import zipfile

# --- CONFIG ---
tile = "p104r074"
data = "fc"

src_dir = Path(f"/home/jovyan/work-easi-eds/data/compat/files/{data}/{tile}")
out_dir = Path("/home/jovyan/work-easi-eds/exports/zips")
zip_name = f"{data}_{tile}_dljmz_files.zip"
zip_path = out_dir / zip_name

# --- CHECK SOURCE ---
if not src_dir.exists():
    raise FileNotFoundError(f"Source directory does not exist:\n{src_dir}")

# --- FIND FILES ---
matches = sorted(
    p for p in src_dir.iterdir()
    if p.is_file() and "dljmz" in p.name.lower()
)

if not matches:
    raise RuntimeError("No files containing 'dljmz' were found.")

# --- ENSURE OUTPUT DIR ---
out_dir.mkdir(parents=True, exist_ok=True)

# --- ZIP FILES ---
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    for p in matches:
        zf.write(p, arcname=p.name)

# --- OUTPUT ---
print(f"[OK] Zipped {len(matches)} files")
print(f"[OK] ZIP created at:\n{zip_path}")
```
