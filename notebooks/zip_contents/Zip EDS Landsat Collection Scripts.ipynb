# Zip EDS Landsat Collection Scripts

This notebook cell packages all Python (`.py`) scripts from the  
`eds_lsat_collection` directory into a single ZIP archive for easy download
from the EASI Jupyter environment.

The ZIP is intended for:
- Manual download from Jupyter
- Archiving or handover
- Upload to external storage (e.g. S3)
- Reproducibility and audit purposes

Only Python source files are included; no data or compiled artifacts are added.


```python
from pathlib import Path
import zipfile
import os

# --- CONFIG ---
# Absolute path is safest on EASI
src_dir = Path("/home/jovyan/work-easi-eds/scripts/easi-scripts/eds_lsat_collection")
zip_path = src_dir.parent / "eds_lsat_collection_scripts.zip"

# --- DEBUG INFO ---
print("Notebook working directory:", Path.cwd())
print("Looking for scripts in:", src_dir)

if not src_dir.exists():
    raise FileNotFoundError(f"Source directory not found:\n  {src_dir}")

py_files = sorted(src_dir.glob("*.py"))

if not py_files:
    raise RuntimeError(f"No .py files found in {src_dir}")

with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    for p in py_files:
        zf.write(p, arcname=p.name)

print(f"[OK] Zipped {len(py_files)} Python scripts")
print(f"[OK] Output ZIP: {zip_path}")
```
